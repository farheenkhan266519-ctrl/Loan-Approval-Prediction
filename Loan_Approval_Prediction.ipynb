```python

# Import the Pandas library
import pandas as pd

# Load the loan approval dataset
df = pd.read_csv("/content/LoanApprovalPrediction.csv")

# Display a success message
print("Dataset loaded successfully!")

# Display the number of rows and columns
print("Dataset Shape:", df.shape)
```

```python

# Count missing values in each column
print("Missing Values Before Cleaning:")

print(df.isnull().sum())
```

```python

# Count duplicate rows in the dataset
print("Duplicate Rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Display the new dataset shape
print("Shape After Removing Duplicates:", df.shape)
```

```python

# Loop through each column in the dataset
for col in df.columns:

    # Check if the column contains text values
    if df[col].dtype == "object":

        # Fill missing values with the most common value
        df[col] = df[col].fillna(df[col].mode()[0])

    # Otherwise, treat the column as numerical
    else:

        # Fill missing values with the median
        df[col] = df[col].fillna(df[col].median())

# Display missing values after cleaning
print("Missing Values After Cleaning:")

print(df.isnull().sum())
```

```python

# Convert categorical text values into numerical columns
df = pd.get_dummies(df, drop_first=True, dtype=int)

# Display the first five rows of the encoded dataset
print("Encoded Dataset:")

display(df.head())
```

```python

# Display the final number of rows and columns
print("Final Dataset Shape:", df.shape)

# Display all column names
print("\nFinal Columns:")

print(df.columns.tolist())

# Display the data type of each column
print("\nData Types:")

print(df.dtypes)
```

```python

# Save the cleaned dataset as a CSV file
df.to_csv("cleaned_Loan_Approval_Prediction.csv", index=False)

# Display a confirmation message
print("Cleaned dataset saved successfully!")
```

```python
import pandas as pd

# Load the prepared dataset
df = pd.read_csv("/content/cleaned_loan_dataset.csv")

# Display dataset information
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

# Show first 5 rows
df.head()
```

```python
# Select the target column
if "Loan_Status" in df.columns:
    target_column = "Loan_Status"
elif "Loan_Status_Y" in df.columns:
    target_column = "Loan_Status_Y"
else:
    raise ValueError("Loan_Status target column not found!")

# Separate input features (X) and target (y)
X = df.drop(columns=[target_column])
y = df[target_column]

# Display feature and target information
print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Target column:", target_column)

print("\nTarget values:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

# Split dataset into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# Create a pipeline with scaling and Logistic Regression
model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)

# Train the model
model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Predict loan approval status for test data
y_pred = model.predict(X_test)

# Display actual and predicted values
results = pd.DataFrame({
    "Actual Loan Status": y_test.values,
    "Predicted Loan Status": y_pred
})

results.head(10)
```

```python
# Get prediction probabilities
probabilities = model.predict_proba(X_test)

# Create a probability table
probability_results = pd.DataFrame(
    probabilities,
    columns=model.classes_
)

probability_results["Actual Loan Status"] = y_test.values
probability_results["Predicted Loan Status"] = y_pred

probability_results.head(10)
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Calculate accuracy on training and testing data
train_accuracy = model.score(X_train, y_train)
test_accuracy = accuracy_score(y_test, y_pred)

print("Training Accuracy:", round(train_accuracy * 100, 2), "%")
print("Testing Accuracy:", round(test_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
# Import required evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

print("Evaluation libraries imported successfully!")
```

```python
# Generate predictions from the trained model
y_pred = model.predict(X_test)

# Identify the positive class
classes = model.classes_
positive_class = 1 if 1 in classes else classes[-1]

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test, y_pred,
    pos_label=positive_class,
    zero_division=0
)

recall = recall_score(
    y_test, y_pred,
    pos_label=positive_class,
    zero_division=0
)

f1 = f1_score(
    y_test, y_pred,
    pos_label=positive_class,
    zero_division=0
)

# Display all metrics
print("LOAN APPROVAL MODEL EVALUATION")
print("-" * 40)
print(f"Accuracy:  {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
```

```python
# Create confusion matrix
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=classes
)

# Display matrix as a graph
display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=classes
)

display.plot(cmap="Blues", values_format="d")
plt.title("Loan Approval - Confusion Matrix")
plt.show()

print("Confusion Matrix:")
print(cm)
```

```python
# Generate detailed classification report
report = classification_report(
    y_test,
    y_pred,
    zero_division=0
)

print("CLASSIFICATION REPORT")
print("-" * 40)
print(report)
```

```python
# Get probability of the positive class
positive_index = list(classes).index(positive_class)
y_probability = model.predict_proba(X_test)[:, positive_index]

# Convert target labels to binary values for ROC-AUC
y_test_binary = (np.asarray(y_test) == positive_class).astype(int)

# ROC-AUC requires both classes in the test data
if len(np.unique(y_test_binary)) == 2:
    roc_auc = roc_auc_score(y_test_binary, y_probability)

    print(f"ROC-AUC Score: {roc_auc:.4f}")

    # Plot ROC curve
    RocCurveDisplay.from_predictions(
        y_test_binary,
        y_probability
    )

    plt.title("Loan Approval - ROC Curve")
    plt.show()
else:
    roc_auc = None
    print("ROC-AUC cannot be calculated because test data has only one class.")
```

```python
print("MODEL PERFORMANCE ANALYSIS")
print("-" * 40)

print(f"Accuracy: {accuracy * 100:.2f}%")

if precision > recall:
    print("Precision is higher than recall.")
    print("The model has relatively fewer false positives among its positive predictions.")
elif recall > precision:
    print("Recall is higher than precision.")
    print("The model identifies a relatively larger share of actual approved cases.")
else:
    print("Precision and recall are equal or very close.")

if roc_auc is not None:
    print(f"ROC-AUC: {roc_auc:.4f}")

    if roc_auc > 0.5:
        print("The model separates the two classes better than random ranking.")
    elif roc_auc == 0.5:
        print("The ROC-AUC is equal to random ranking.")
    else:
        print("The ROC-AUC is below random ranking; check class labels and model setup.")

print("\nConfusion Matrix:")
print(cm)
```

```python
# Create a summary table of evaluation results
evaluation_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC"
    ],
    "Score": [
        accuracy,
        precision,
        recall,
        f1,
        roc_auc
    ]
})

evaluation_results
```

```python
import pandas as pd

# Create a sample applicant
sample_applicant = {}

for column in X_train.columns:
    sample_applicant[column] = X_train[column].mode().iloc[0]

# Show applicant information
applicant_df = pd.DataFrame([sample_applicant])

print("Applicant Information:")
display(applicant_df)
```

```python
# Predict loan status
prediction = model.predict(applicant_df)[0]

# Get probabilities
probabilities = model.predict_proba(applicant_df)[0]

print("Loan Prediction:", prediction)

print("\nPrediction Probabilities:")

for class_name, probability in zip(
    model.classes_,
    probabilities
):
    print(f"Class {class_name}: {probability * 100:.2f}%")
```

```python
# Create three sample applicants
applicant_1 = sample_applicant.copy()
applicant_2 = sample_applicant.copy()
applicant_3 = sample_applicant.copy()

# Change one numeric feature for different test cases
numeric_columns = X_train.select_dtypes(
    include="number"
).columns.tolist()

if numeric_columns:
    feature = numeric_columns[0]

    applicant_2[feature] = X_train[feature].median()
    applicant_3[feature] = X_train[feature].max()

# Store applicants
applicants = [
    applicant_1,
    applicant_2,
    applicant_3
]

# Predict each applicant
results = []

for i, applicant in enumerate(applicants, start=1):

    applicant_df = pd.DataFrame([applicant])

    prediction = model.predict(applicant_df)[0]
    probabilities = model.predict_proba(applicant_df)[0]

    predicted_index = list(model.classes_).index(prediction)

    results.append({
        "Applicant": f"Applicant {i}",
        "Prediction": prediction,
        "Probability (%)": round(
            probabilities[predicted_index] * 100, 2
        )
    })

# Show results
results_df = pd.DataFrame(results)
display(results_df)
```
